```python
import time
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

def scrape_disease(url):

    driver.get(url)

    try:
        WebDriverWait(driver, 15).until(
            EC.presence_of_element_located((By.TAG_NAME, "body"))
        )
    except:
        return None

    time.sleep(1)

    disease_data = {}

    try:
        disease_name = driver.find_element(
            By.TAG_NAME, "h1"
        ).text.strip()
    except:
        return None

    if not disease_name:
        return None

    disease_data["Disease"] = disease_name

    elements = driver.find_elements(
        By.XPATH,
        "//h2 | //h3 | //p"
    )

    current_title = "Introduction"
    disease_data[current_title] = ""

    for element in elements:

        tag = element.tag_name
        text = element.text.strip()

        if not text:
            continue

        if text in [
            "NHS inform",
            "Also on NHS inform",
            "Other health sites",
            "Find your nearest pharmacy"
        ]:
            continue

        if "©" in text:
            continue

        if tag in ["h2", "h3"]:

            current_title = text

            if current_title not in disease_data:
                disease_data[current_title] = ""

        elif tag == "p":

            if current_title not in disease_data:
                disease_data[current_title] = ""

            disease_data[current_title] += text + "\n"

    for key in disease_data:
        disease_data[key] = disease_data[key].strip()

    return disease_data
```

```python
import time
import json

from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC


COOKIE_TITLE = "Your choice regarding cookies on this site"

GENERIC_TITLES = {
    "Illnesses and conditions",
    "About NHS inform",
    "Home",
    "Cookie settings",
    "Cookies on NHS inform",
    COOKIE_TITLE
}


def accept_cookies():

    xpaths = [
        "//button[contains(translate(., 'ABCDEFGHIJKLMNOPQRSTUVWXYZ', 'abcdefghijklmnopqrstuvwxyz'), 'accept all')]",
        "//button[contains(translate(., 'ABCDEFGHIJKLMNOPQRSTUVWXYZ', 'abcdefghijklmnopqrstuvwxyz'), 'accept cookies')]",
        "//button[contains(translate(., 'ABCDEFGHIJKLMNOPQRSTUVWXYZ', 'abcdefghijklmnopqrstuvwxyz'), 'allow all')]",
        "//a[contains(translate(., 'ABCDEFGHIJKLMNOPQRSTUVWXYZ', 'abcdefghijklmnopqrstuvwxyz'), 'accept all')]"
    ]

    for xpath in xpaths:

        try:

            button = WebDriverWait(driver, 3).until(
                EC.element_to_be_clickable((By.XPATH, xpath))
            )

            driver.execute_script(
                "arguments[0].click();",
                button
            )

            time.sleep(1)

            return True

        except:
            pass

    return False


def get_h1():

    try:

        element = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.TAG_NAME, "h1"))
        )

        return element.text.strip()

    except:

        return None


def scrape_disease(url):

    driver.get(url)

    time.sleep(3)

    accept_cookies()

    time.sleep(1)

    disease_name = get_h1()

    if not disease_name:

        raise Exception("Could not extract disease name")

    if disease_name == COOKIE_TITLE:

        accept_cookies()

        time.sleep(2)

        disease_name = get_h1()

    if not disease_name:

        raise Exception("Could not extract disease name after cookie handling")

    if disease_name in GENERIC_TITLES:

        raise Exception(
            f"Invalid page title: {disease_name}"
        )

    elements = driver.find_elements(
        By.XPATH,
        "//h2 | //h3 | //p"
    )

    if len(elements) < 2:

        raise Exception(
            "Page does not contain enough content"
        )

    disease_data = {}

    disease_data["Disease"] = disease_name

    current_section = "Introduction"

    disease_data[current_section] = ""

    for element in elements:

        tag = element.tag_name

        text = element.text.strip()

        if not text:

            continue

        if tag in ["h2", "h3"]:

            current_section = text

            if current_section not in disease_data:

                disease_data[current_section] = ""

        elif tag == "p":

            if current_section not in disease_data:

                disease_data[current_section] = text

            elif disease_data[current_section]:

                disease_data[current_section] += "\n" + text

            else:

                disease_data[current_section] = text

    content_sections = [
        key for key in disease_data
        if key != "Disease" and disease_data[key].strip()
    ]

    if len(content_sections) == 0:

        raise Exception(
            "No useful disease content found"
        )

    return disease_data


all_diseases_final = []

failed_pages_final = []

total = len(valid_links)

print("=" * 70)
print("FINAL MEDICAL WEB SCRAPING")
print("=" * 70)
print("Total valid links:", total)
print()


for i, (name, url) in enumerate(valid_links, start=1):

    print(f"[{i}/{total}] {name}")

    success = False

    last_error = ""

    for attempt in range(1, 4):

        try:

            data = scrape_disease(url)

            all_diseases_final.append(data)

            print(
                f"    SUCCESS - {data['Disease']}"
            )

            success = True

            break

        except Exception as e:

            last_error = str(e)

            print(
                f"    Attempt {attempt} failed: {last_error}"
            )

            time.sleep(2)

    if not success:

        failed_pages_final.append({
            "Name": name,
            "URL": url,
            "Error": last_error
        })

        print("    FAILED")


print()
print("=" * 70)
print("SCRAPING FINISHED")
print("=" * 70)

print("Total valid links:", total)

print(
    "Successful:",
    len(all_diseases_final)
)

print(
    "Failed:",
    len(failed_pages_final)
)


cookie_pages = []

generic_pages = []

empty_pages = []


for disease in all_diseases_final:

    disease_name = disease.get(
        "Disease",
        ""
    ).strip()

    if disease_name == COOKIE_TITLE:

        cookie_pages.append(disease)

    if disease_name in GENERIC_TITLES:

        generic_pages.append(disease)

    content_count = 0

    for key, value in disease.items():

        if key != "Disease" and isinstance(value, str):

            if value.strip():

                content_count += 1

    if content_count == 0:

        empty_pages.append(disease)


print()
print("=" * 70)
print("VALIDATION")
print("=" * 70)

print(
    "Cookie pages:",
    len(cookie_pages)
)

print(
    "Generic pages:",
    len(generic_pages)
)

print(
    "Empty pages:",
    len(empty_pages)
)


print()
print("First 10 collected pages:")

for i, disease in enumerate(
    all_diseases_final[:10],
    start=1
):

    print(
        i,
        "->",
        disease.get("Disease")
    )


with open(
    "medical_diseases_final.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        all_diseases_final,
        file,
        ensure_ascii=False,
        indent=4
    )


with open(
    "failed_pages_final.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        failed_pages_final,
        file,
        ensure_ascii=False,
        indent=4
    )


print()
print("=" * 70)
print("FILES SAVED")
print("=" * 70)

print("medical_diseases_final.json")
print("failed_pages_final.json")
```

```python
import json
import re


# =========================================================
# 1. Load the original dataset
# =========================================================

with open(
    "medical_diseases_complete.json",
    "r",
    encoding="utf-8"
) as file:

    diseases = json.load(file)


print("Original records:", len(diseases))


# =========================================================
# 2. Sections that are navigation/footer information
# =========================================================

remove_sections = {
    "NHS inform",
    "Other health sites",
    "Also on NHS inform",
    "Find your nearest sexual health service"
}


# =========================================================
# 3. Text cleaning function
# =========================================================

def clean_text(text):

    if not isinstance(text, str):
        return text

    # Remove repeated spaces
    text = re.sub(r"[ \t]+", " ", text)

    # Remove excessive blank lines
    text = re.sub(r"\n\s*\n+", "\n\n", text)

    # Remove spaces at beginning/end of lines
    text = "\n".join(
        line.strip()
        for line in text.splitlines()
    )

    return text.strip()


# =========================================================
# 4. Clean every disease
# =========================================================

cleaned_diseases = []

removed_sections_count = 0
empty_sections_count = 0


for disease in diseases:

    cleaned = {}

    for section, content in disease.items():

        section = clean_text(section)

        # Keep the disease name
        if section == "Disease":

            cleaned["Disease"] = clean_text(content)
            continue


        # Remove known navigation/footer sections
        if section in remove_sections:

            removed_sections_count += 1
            continue


        # Clean the actual content
        cleaned_content = clean_text(content)


        # Skip empty sections
        if not cleaned_content:

            empty_sections_count += 1
            continue


        cleaned[section] = cleaned_content


    cleaned_diseases.append(cleaned)


# =========================================================
# 5. Save cleaned dataset
# =========================================================

with open(
    "medical_diseases_cleaned.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        cleaned_diseases,
        file,
        ensure_ascii=False,
        indent=4
    )


# =========================================================
# 6. Cleaning report
# =========================================================

print()
print("=" * 70)
print("CLEANING FINISHED")
print("=" * 70)

print("Original records:", len(diseases))
print("Cleaned records:", len(cleaned_diseases))

print(
    "Removed navigation/footer sections:",
    removed_sections_count
)

print(
    "Removed empty sections:",
    empty_sections_count
)

print()
print("Output file:")
print("medical_diseases_cleaned.json")
```

```python
import json
import re


# =========================================================
# 1. Load cleaned dataset
# =========================================================

with open(
    "medical_diseases_cleaned.json",
    "r",
    encoding="utf-8"
) as file:

    cleaned_diseases = json.load(file)


# =========================================================
# 2. Chunk settings
# =========================================================

CHUNK_SIZE = 1000
CHUNK_OVERLAP = 150

smart_chunks = []


# =========================================================
# 3. Function to find a good place to cut
# =========================================================

def find_cut_position(text, start, target_end):

    # Don't search before the beginning
    search_start = start + 300

    # Don't go beyond the text
    target_end = min(target_end, len(text))

    # -----------------------------------------
    # Try to find the last sentence ending
    # -----------------------------------------

    sentence_endings = [
        ". ",
        "! ",
        "? ",
        ".\n",
        "!\n",
        "?\n"
    ]

    best_position = -1

    for ending in sentence_endings:

        position = text.rfind(
            ending,
            search_start,
            target_end
        )

        if position > best_position:
            best_position = position + len(ending)

    if best_position != -1:
        return best_position


    # -----------------------------------------
    # If no sentence ending, cut at last space
    # -----------------------------------------

    space_position = text.rfind(
        " ",
        search_start,
        target_end
    )

    if space_position != -1:
        return space_position


    # -----------------------------------------
    # Last fallback
    # -----------------------------------------

    return target_end


# =========================================================
# 4. Create smart chunks
# =========================================================

for disease in cleaned_diseases:

    disease_name = disease["Disease"]

    for section, content in disease.items():

        if section == "Disease":
            continue

        text = content.strip()

        if not text:
            continue

        start = 0

        while start < len(text):

            target_end = min(
                start + CHUNK_SIZE,
                len(text)
            )

            # If the remaining text is already small,
            # keep it as the final chunk.
            if len(text) - start <= CHUNK_SIZE:

                chunk_text = text[start:].strip()

                if chunk_text:
                    smart_chunks.append({
                        "Disease": disease_name,
                        "Section": section,
                        "Text": chunk_text
                    })

                break


            # Find a natural cutting point
            cut_position = find_cut_position(
                text,
                start,
                target_end
            )

            chunk_text = text[start:cut_position].strip()

            if chunk_text:

                smart_chunks.append({
                    "Disease": disease_name,
                    "Section": section,
                    "Text": chunk_text
                })

            # Move forward while keeping overlap
            start = max(
                cut_position - CHUNK_OVERLAP,
                start + 1
            )


# =========================================================
# 5. Statistics
# =========================================================

chunk_lengths = [
    len(chunk["Text"])
    for chunk in smart_chunks
]


print("=" * 70)
print("SMART CHUNKING FINISHED")
print("=" * 70)

print("Total diseases:", len(cleaned_diseases))
print("Total chunks:", len(smart_chunks))

print(
    "Average chunk characters:",
    round(
        sum(chunk_lengths) / len(chunk_lengths)
    )
)

print(
    "Minimum chunk characters:",
    min(chunk_lengths)
)

print(
    "Maximum chunk characters:",
    max(chunk_lengths)
)


# =========================================================
# 6. Show first 5 chunks
# =========================================================

print()
print("=" * 70)
print("FIRST 5 CHUNKS")
print("=" * 70)

for i, chunk in enumerate(
    smart_chunks[:5],
    start=1
):

    print()
    print("-" * 70)
    print("Chunk:", i)
    print("Disease:", chunk["Disease"])
    print("Section:", chunk["Section"])
    print("Characters:", len(chunk["Text"]))
    print("Text:")
    print(chunk["Text"][:500])


# =========================================================
# 7. Save smart chunks
# =========================================================

with open(
    "medical_chunks.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        smart_chunks,
        file,
        ensure_ascii=False,
        indent=4
    )


print()
print("=" * 70)
print("Saved:")
print("medical_chunks.json")
print("=" * 70)
```

```python
import json


# =========================================================
# 1. Load smart chunks
# =========================================================

with open(
    "medical_chunks.json",
    "r",
    encoding="utf-8"
) as file:

    smart_chunks = json.load(file)


MIN_CHUNK_SIZE = 200

final_chunks = []


# =========================================================
# 2. Merge very small chunks
# =========================================================

for chunk in smart_chunks:

    text = chunk["Text"].strip()

    # Chunk is large enough
    if len(text) >= MIN_CHUNK_SIZE:

        final_chunks.append({
            "Disease": chunk["Disease"],
            "Section": chunk["Section"],
            "Text": text
        })

        continue


    # Very small chunk
    # Try to merge it with the previous chunk
    if final_chunks:

        previous = final_chunks[-1]

        same_disease = (
            previous["Disease"] == chunk["Disease"]
        )

        same_section = (
            previous["Section"] == chunk["Section"]
        )

        if same_disease and same_section:

            previous["Text"] += " " + text
            continue


    # If we cannot merge it safely,
    # keep it instead of deleting medical information
    final_chunks.append({
        "Disease": chunk["Disease"],
        "Section": chunk["Section"],
        "Text": text
    })


# =========================================================
# 3. Statistics
# =========================================================

lengths = [
    len(chunk["Text"])
    for chunk in final_chunks
]


print("=" * 70)
print("FINAL CHUNKING")
print("=" * 70)

print("Total diseases:", len(set(
    chunk["Disease"]
    for chunk in final_chunks
)))

print("Total chunks:", len(final_chunks))

print(
    "Average chunk characters:",
    round(sum(lengths) / len(lengths))
)

print(
    "Minimum chunk characters:",
    min(lengths)
)

print(
    "Maximum chunk characters:",
    max(lengths)
)


# =========================================================
# 4. Count very small chunks
# =========================================================

small_chunks = [
    chunk
    for chunk in final_chunks
    if len(chunk["Text"]) < MIN_CHUNK_SIZE
]


print(
    "Chunks below",
    MIN_CHUNK_SIZE,
    "characters:",
    len(small_chunks)
)


# =========================================================
# 5. Save final chunks
# =========================================================

with open(
    "medical_chunks_final.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        final_chunks,
        file,
        ensure_ascii=False,
        indent=4
    )


print()
print("Saved:")
print("medical_chunks_final.json")
```

```python
import json
import re


# =========================================================
# 1. Load cleaned medical dataset
# =========================================================

with open(
    "medical_diseases_cleaned.json",
    "r",
    encoding="utf-8"
) as file:

    diseases = json.load(file)


CHUNK_SIZE = 1000
CHUNK_OVERLAP = 150
MIN_CHUNK_SIZE = 200


# =========================================================
# 2. Prepare text for each disease
# =========================================================

prepared_documents = []


for disease in diseases:

    disease_name = disease["Disease"]

    sections = []

    for section, content in disease.items():

        if section == "Disease":
            continue

        content = str(content).strip()

        if not content:
            continue

        # Keep the section name inside the text
        section_text = (
            f"[{section}]\n"
            f"{content}"
        )

        sections.append(section_text)


    full_text = "\n\n".join(sections)

    prepared_documents.append({
        "Disease": disease_name,
        "Text": full_text
    })


# =========================================================
# 3. Find a natural cutting point
# =========================================================

def find_cut_position(text, start, target_end):

    search_start = start + 300

    target_end = min(
        target_end,
        len(text)
    )

    # Prefer sentence endings
    endings = [
        ". ",
        "! ",
        "? ",
        ".\n",
        "!\n",
        "?\n"
    ]

    best_position = -1

    for ending in endings:

        position = text.rfind(
            ending,
            search_start,
            target_end
        )

        if position > best_position:
            best_position = position + len(ending)


    if best_position != -1:
        return best_position


    # Otherwise cut at the last space
    space_position = text.rfind(
        " ",
        search_start,
        target_end
    )

    if space_position != -1:
        return space_position


    return target_end


# =========================================================
# 4. Create final chunks
# =========================================================

final_chunks = []


for document in prepared_documents:

    disease_name = document["Disease"]
    text = document["Text"]

    start = 0

    while start < len(text):

        remaining = len(text) - start


        # ---------------------------------------------
        # Last part of the document
        # ---------------------------------------------

        if remaining <= CHUNK_SIZE:

            chunk_text = text[start:].strip()

            if chunk_text:
                final_chunks.append({
                    "Disease": disease_name,
                    "Text": chunk_text
                })

            break


        # ---------------------------------------------
        # Find natural cutting point
        # ---------------------------------------------

        cut_position = find_cut_position(
            text,
            start,
            start + CHUNK_SIZE
        )

        chunk_text = text[
            start:cut_position
        ].strip()


        if chunk_text:

            final_chunks.append({
                "Disease": disease_name,
                "Text": chunk_text
            })


        # ---------------------------------------------
        # Move forward with overlap
        # ---------------------------------------------

        next_start = cut_position - CHUNK_OVERLAP

        if next_start <= start:
            next_start = cut_position

        start = next_start


# =========================================================
# 5. Remove ONLY accidental tiny fragments
# =========================================================

clean_chunks = []

small_fragments = []


for chunk in final_chunks:

    text = chunk["Text"].strip()

    if len(text) >= MIN_CHUNK_SIZE:

        clean_chunks.append(chunk)

    else:

        small_fragments.append(chunk)


# =========================================================
# 6. Safely merge remaining small fragments
# =========================================================

for fragment in small_fragments:

    merged = False

    # Try to merge with nearest chunk from same disease
    for i in range(len(clean_chunks) - 1, -1, -1):

        if clean_chunks[i]["Disease"] == fragment["Disease"]:

            combined = (
                clean_chunks[i]["Text"]
                + "\n\n"
                + fragment["Text"]
            )

            # Don't make a huge chunk
            if len(combined) <= CHUNK_SIZE + 300:

                clean_chunks[i]["Text"] = combined

                merged = True
                break


    # If no safe merge exists, keep it.
    # We never delete medical information.
    if not merged:

        clean_chunks.append(fragment)


# =========================================================
# 7. Final statistics
# =========================================================

lengths = [
    len(chunk["Text"])
    for chunk in clean_chunks
]


small_chunks = [
    chunk
    for chunk in clean_chunks
    if len(chunk["Text"]) < MIN_CHUNK_SIZE
]


print("=" * 70)
print("FINAL SMART CHUNKING")
print("=" * 70)

print("Total diseases:", len(diseases))

print("Total chunks:", len(clean_chunks))

print(
    "Average chunk characters:",
    round(sum(lengths) / len(lengths))
)

print(
    "Minimum chunk characters:",
    min(lengths)
)

print(
    "Maximum chunk characters:",
    max(lengths)
)

print(
    "Chunks below",
    MIN_CHUNK_SIZE,
    "characters:",
    len(small_chunks)
)


# =========================================================
# 8. Show first 5 chunks
# =========================================================

print()
print("=" * 70)
print("FIRST 5 FINAL CHUNKS")
print("=" * 70)

for i, chunk in enumerate(
    clean_chunks[:5],
    start=1
):

    print()
    print("-" * 70)

    print("Chunk:", i)
    print("Disease:", chunk["Disease"])
    print("Characters:", len(chunk["Text"]))

    print(
        chunk["Text"][:700]
    )


# =========================================================
# 9. Save final dataset
# =========================================================

with open(
    "medical_chunks_final.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        clean_chunks,
        file,
        ensure_ascii=False,
        indent=4
    )


print()
print("=" * 70)
print("Saved:")
print("medical_chunks_final.json")
print("=" * 70)
```

```python
import json
import numpy as np


# =========================================================
# 1. Load final chunks
# =========================================================

with open(
    "medical_chunks_final.json",
    "r",
    encoding="utf-8"
) as file:

    chunks = json.load(file)


print("Total chunks:", len(chunks))


# =========================================================
# 2. Extract text
# =========================================================

texts = [
    chunk["Text"]
    for chunk in chunks
]


# =========================================================
# 3. Create embeddings
# =========================================================

embeddings = model.encode(
    texts,
    batch_size=32,
    show_progress_bar=True,
    normalize_embeddings=True
)


# =========================================================
# 4. Verify embeddings
# =========================================================

print()
print("=" * 60)
print("EMBEDDINGS FINISHED")
print("=" * 60)

print("Total chunks:", len(chunks))
print("Total embeddings:", len(embeddings))
print("Embedding dimensions:", embeddings.shape[1])


# =========================================================
# 5. Verify count
# =========================================================

if len(chunks) == len(embeddings):

    print("SUCCESS: Every chunk has an embedding!")

else:

    print("WARNING: Number of chunks and embeddings do not match!")
```

```python
from pymongo import MongoClient, UpdateOne


# =========================================================
# 1. Connect to MongoDB
# =========================================================

client = MongoClient("mongodb://localhost:27017/")

db = client["medical_ai"]

collection = db["medical_chunks"]


# =========================================================
# 2. Prepare bulk updates
# =========================================================

operations = []


for i, chunk in enumerate(chunks):

    operations.append(
        UpdateOne(
            {
                "Disease": chunk["Disease"],
                "Text": chunk["Text"]
            },
            {
                "$set": {
                    "embedding": embeddings[i].tolist()
                }
            }
        )
    )


# =========================================================
# 3. Update MongoDB in batches
# =========================================================

BATCH_SIZE = 500

total_updated = 0


for start in range(
    0,
    len(operations),
    BATCH_SIZE
):

    batch = operations[
        start:start + BATCH_SIZE
    ]

    result = collection.bulk_write(
        batch,
        ordered=False
    )

    total_updated += result.modified_count

    print(
        f"Processed: "
        f"{min(start + BATCH_SIZE, len(operations))}"
        f"/{len(operations)}"
    )


# =========================================================
# 4. Verify
# =========================================================

documents_with_embeddings = collection.count_documents(
    {
        "embedding": {
            "$exists": True
        }
    }
)


print()
print("=" * 60)
print("EMBEDDINGS SAVED TO MONGODB")
print("=" * 60)

print(
    "Documents with embeddings:",
    documents_with_embeddings
)

print(
    "Expected:",
    len(chunks)
)


if documents_with_embeddings == len(chunks):

    print(
        "SUCCESS: All embeddings are stored correctly!"
    )

else:

    print(
        "WARNING: Some embeddings are missing!"
    )
```
